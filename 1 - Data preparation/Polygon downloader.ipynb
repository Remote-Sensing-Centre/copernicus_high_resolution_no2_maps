```python
import osmnx as ox
import matplotlib.pyplot as plt
```

## Settings

Go to the [OpenStreetMap](https://www.openstreetmap.org) and search for the region name you are interested in.
Click on the appropriate relation link and copy it's ID. In the image below ID of the Warsaw relation is 336075.

![osm_warsaw.png](img/osm_warsaw.png)

```python
LABEL = "lubuskie"
OSM_ID = 130969
```

## Download and save

```python
gdf = ox.geocode_to_gdf(f"R{OSM_ID}", by_osmid=True)
gdf.to_file(f"{LABEL}_{OSM_ID}.geojson", driver="GeoJSON")
```

## Plot polygon

```python
gdf.plot(figsize=(8, 8), edgecolor="black", facecolor="lightgray")
plt.title(f"OSM Polygon for ID {OSM_ID}")
plt.show()
```
